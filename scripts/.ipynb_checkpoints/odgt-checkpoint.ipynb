```python
import json

def read_odgt(path):
    records = []
    with open(path, "r") as f:
        for line in f:
            if line.strip():
                records.append(json.loads(line))
    return records
```

```python
data = read_odgt("../../CrowdHuman/annotation_train.odgt")
print(len(data))
print(data[0].keys())
```

```python
def xywh_to_xyxy(box):
    x, y, w, h = box
    return [x, y, x + w, y + h]
```

```python
import torch

def parse_annotation(record):
    boxes = []
    labels = []

    for obj in record["gtboxes"]:
        if obj["head_attr"]["ignore"] == 1:
            continue

        box = xywh_to_xyxy(obj["fbox"])
        boxes.append(box)
        labels.append(1)  # person class

    boxes = torch.tensor(boxes, dtype=torch.float32)
    labels = torch.tensor(labels, dtype=torch.int64)

    return boxes, labels
```

```python
boxes, labels = parse_annotation(data[0])
print(boxes)
print(labels)
```

```python
def build_target(record, idx):
    boxes, labels = parse_annotation(record)

    target = {
        "boxes": boxes,
        "labels": labels,
        "image_id": torch.tensor([idx])
    }

    return target
```

```python
from torch.utils.data import Dataset

class CrowdHumanDataset(Dataset):
    def __init__(self, odgt_path, image_dir):
        self.records = read_odgt(odgt_path)
        self.image_dir = image_dir

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        record = self.records[idx]

        image = load_image(self.image_dir, record["ID"])
        target = build_target(record, idx)

        return image, target
```

```python
from PIL import Image
import torchvision.transforms as T
import os

_transform = T.ToTensor()

def load_image(image_dir, image_id):
    img_path = os.path.join(image_dir, image_id + ".jpg")
    image = Image.open(img_path).convert("RGB")
    return _transform(image)
```

```python
dataset = CrowdHumanDataset(
    odgt_path="../../CrowdHuman/annotation_train.odgt",
    image_dir="../../CrowdHuman/images/train"
)

img, target = dataset[0]

print(img.shape)        # [3, H, W]
print(target["boxes"])  # [N, 4]
print(target["labels"]) # [N]
```

```python
from torchvision.models.detection import fasterrcnn_resnet50_fpn

model = fasterrcnn_resnet50_fpn(num_classes=2)  # background + person

images, targets = [img], [target]
losses = model(images, targets)

print(losses)
```
